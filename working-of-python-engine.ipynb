# Working of Python Engine

## Introduction: What is Python?

First published by programmer Guido van Rossum in February 1991, Python is a popular and versatile programming language. It has been used for a long time in web development, scientific computation and automation, and its even higher popularity in recent years can be correlated to people's growing interest in Artificial Intelligence and Machine Learning.

What makes Python easy to learn is its high-level nature, which means it has a high level of abstraction. Think of abstraction as low to high layers that "cover" the engines of a system. In Python, we write code at a high level of abstraction, or a simplified layer of the language, meaning we are not seeing the internal workings, only its outputs.

Despite not seeing the inner workings of the language, we still deal with processes like variable assignment, code interpretation and code execution. Now, this article will help you to understand what's behind your Python code and why it works!

## How Python Code Is Executed: from source code to the Python Virtual Machine

To put it simply, the execution of Python code starts with source code, the written instructions in a programming language before any execution, like this:

```python
number1 = 10
number2 = 5
result = number1 + number2
print(result)
```

Then this source code will be processed by CPython, an implementation of Python in the C programming language.

Firstly, CPython parses the source code, checking whether the grammar rules of Python are followed. If successful, the process continues and the code is compiled into bytecode, an intermediate representation of the code. Now, the CPython interpreter can decode the bytecode and execute the corresponding operations, respecting what is defined by the PVM (Python Virtual Machine), a conceptual model that defines the execution of Python bytecode.

The cell below demonstrates how to use the dis module. It allows us to inspect the bytecode generated by the compiler, displaying the instructions in a human-readable format:

```python
import dis

def add_numbers():
    number1 = 10
    number2 = 5
    result = number1 + number2
    return result

dis.dis(add_numbers)
```

These steps provide an intuitive way of understanding the sequential execution of instructions. LOAD_CONST loads the value 10 onto the stack, and STORE_FAST stores it in the space associated with the variable number1. The same happens with the value 5 and the variable number2. In LOAD_FAST_LOAD_FAST, both number1 and number2 are loaded onto the stack for the next operation. Then, BINARY_OP performs the addition (+), taking both values from the stack and placing their sum (15) back onto it. Finally, STORE_FAST takes 15 from the stack and stores it in the variable result, while RETURN_VALUE returns it at the end of the function.

## Conclusion

In conclusion, Python's high-level nature makes programming more accessible by hiding much of the complexity behind code execution. However, understanding what happens behind the scenes helps us recognize how different components, such as CPython, bytecode and the Python Virtual Machine, work together to make our code run.
