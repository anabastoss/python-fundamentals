# The Python Syxtax

## What is syntax?

In programming, syntax refers to the set of rules that defines how instructions must be written. In Python, following these rules allows the interpreter to understand and execute our code correctly.

## Variables and Assignment

Variables can be compared to labels attached to drawers on a shelf. They give us a way to identify and access values stored in memory. To assign a value to a variable, we can perform the following code:

```python
variable = 'value'
a = 1
b = 2.5
```

The correct assignment operator in Python is =, not == (equality operator) or : (commonly used in dictionaries).

## Basic Data Types

There are four main data types or classes in Python:

- Strings (str)
- Integers (int)
- Floats (float)
- Booleans (bool)

A string is a data type used to store sequences of characters, such as letters, numbers and symbols. They represent text, and can be defined by using quotation marks, either single or double quotes:

```python
c = 'string'
d = "letters"
```

Integers represent whole numbers, while floating-point numbers represent real-number approximations using decimal notation:

```python
e = 1       # an integer
f = 4.5     # an float
```

Unlike strings, they don't require quotation marks. Actually, if you try to create numbers with quotes, they become strings:

```python
print(type('2.5'))
print(type('1'))
# These values belong to the string class
```

Booleans are logical data types that represent two possible truth values: True and False. In Python, True and False are also associated with the integer values 1 and 0, respectively. A common use for them is to test conditions (which I am not going to explain by now).

```python
print(f'True is {bool(True)}')
print(f'False is {bool(False)}')
print(f'1 is {bool(1)}')
print(f'"0" is {bool("0")}')
print(f'0 is {bool(0)}')
print(f'Empty string is {bool('')}')
print(f'None is {bool(None)}')
```

The bool() function evaluates the truth value of different objects. Notice how even though "0" looks like zero, it is considered True because it is a non-empty string!

## Indentation and Comments

Indentation is the use of spaces at the beginning of a code line. It defines blocks of code and determines which instructions belong together. You can use the Tab key, but it is not recommended to mix tabs and spaces, as they may be interpreted differently. Instead, the most common and recommended convention is to use four spaces per indentation level.

On the other hand, comments and docstrings help programmers explain the logic and purpose of their code. Python provides different ways to add these explanations to the source code.

Single-line comments begin with the hash symbol (#). They are ignored during code execution and can also be written after an instruction:

```python
felidae = 'cat' # cats are part of felidae
```

Docstrings are strings used to document the purpose and behavior of modules, functions and classes. They are written using triple quotation marks and placed immediately after the definition of the object being documented.

```python
"""
This program introduces basic Python syntax.
It demonstrates variables and data types.
"""

name = "Joao"
age = 40
```
